- Create a subclass `SavingsAccount` that adds an `interest_rate` and a method `add_interest()` which increases the balance.

````python
#base class BankAccount with parameters: owner and balance with value 0
class BankAccount():
    def __init__(self,owner,balance=0):
        self.owner = owner
        self.balance = balance
#method deposit with amount parameter 
    def deposit(self,amount):
        self.balance += amount
        print(f"Deposited N{amount} successfullly. New Balance: N{self.balance}")
#method withdraw with amount parameter and an if statement so users can't withdraw below the funds available in their account.
    def withdraw(self,amount):
        if amount > self.balance:
            print("Insufficient funds")
        else:
            self.balance -= amount
            print(f"Withdrew N{amount} successfully. New Balance: N{self.balance}")

#subclass SavingsAccount that adds an interest_rate
class SavingsAccount(BankAccount):
    def __init__(self, owner, balance=0,interest_rate=0.01):
        super().__init__(owner, balance)
        self.interest_rate = interest_rate

#Method Add_interest which increases the balance
    def add_Interest(self):
        interest = self.balance * self.interest_rate
        self.balance += interest
        print(f"Interest added: N{interest}. New Balance: N{self.balance}")

#a savings account for Divine with N2000 balance and 0.05 interest
account1 = SavingsAccount("Divine", 2000, 0.05)
account1.deposit(4000)
account1.withdraw(1000)
account1.add_Interest()
````

- Create a class `CountDown` that takes a number `n` and counts down to 1.  
- Implement the iterator protocol (`__iter__()` and `__next__()` methods).  
- Use it in a `for` loop to print numbers in reverse order.

````python
#class countDown 
class CountDown:
    def __init__(self,n):
        self.n = n #saves the original starting number
        self.number = n  #keeps track of the current countdown number

    def __iter__(self):
        return self

#implements the next value while temporarily saving the current number and 
#decreasing by 1 for the next round but raises StopIteration when the number = 0

    def __next__(self):
        if self.number <= 0:
            raise StopIteration
        x = self.number
        self.number -= 1
        return x

#for loop which prints numbers in reverse order
countdown = CountDown(15)
for num in countdown:
    print(num)
    
````

- Create a class `EvenNumbers` that generates even numbers starting from 0.  
- The iterator should be infinite (no `StopIteration`).  
- Demonstrate usage by printing the first 10 even numbers using a loop and `break`.

````python
#EvenNumbers class which generates even numbers starting from 0
class EvenNumbers:
    def __init__(self,number=0):
        self.number = number

    def __iter__(self):
        return self

    #saves the current number and increment by 2 for the next value
    def __next__(self):
        even = self.number
        self.number += 2
        return even

#instantiating the infinite iterator
evenNo = EvenNumbers()

#loop through and wrap the infinite iterator in enumerate which keep track of the loop
# count (i) starting from 0. When i = 10, the break triggers.
for i, num in enumerate(evenNo):
    if i >= 10:
        break
    print(num)
    
````

- Create a custom iterator `FileLineReader` that takes a filename and reads one line at a time.  
- Implement `__iter__()` and `__next__()` so it can be used in a `for` loop.  
- Test it by reading from a sample `.txt` file and printing each line.

````python
#custom iterator which takes a filename and opens a file in read only mode 
class FileLineReader:
    def __init__(self,filename):
        self.filename = filename
        self.file = open(filename, "r")

    def __iter__(self):
        return self

#reads the file one line at a time
    def __next__(self):
        line = self.file.readline()
        if line:
            return line.strip()  #remove extra spaces if line exists
        else:
            self.file.close()  #closes the file at the end and raises StopIteration
            raise StopIteration

#created a sample text file separating each character with \n
with open("sample.txt", "w") as f:
    f.write("Line 1\n")
    f.write("Line 2\n")
    f.write("Line 3\n")
    f.write("Line 4\n")
    f.write("Line 5\n")

#reading each line with a for loop 
fileReader = FileLineReader("sample.txt")
for line in fileReader:
    print(line)
````

- Create an abstract class `Shape` with an abstract method `area()`.  
- Create subclasses `Rectangle` and `Circle` that implement the `area()` method.  
- Create a list of different shapes and loop through them, printing their area without checking their type.


````python
#abstract class with abstract method area
from abc import ABC, abstractmethod
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

#suclass rectangle that implements area method
class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    def area(self):
        return self.width * self.height

#suclass circle that implements area method
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return 3.14 * (self.radius ** 2)

#a list of different shape and loop through without checking their type while printing
shapeList = [
        Rectangle(5,4),
        Circle(8),
        Rectangle(2,6),
        Circle(4)
]
for shape in shapeList:
    print(f"Area: {shape.area():.2f}")
````

````python
#an iterator class which takes a limit
class EvenNumbers:
    def __init__(self,limit):
        self.limit = limit 
        self.number = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.number <= self.limit:
            num = self.number
            self.number += 2
            return num
        else:
            raise StopIteration

for n in EvenNumbers(20):
    print(n)
````

Create a class `Fibonacci` that generates Fibonacci numbers up to a given `n` using an iterator protocol.

````python
#Fibonnaci sequence where each number is a sum of the two preceding ones
class Fibonacci:
    def __init__(self, n):
        self.n = n  #maximum count of numbers to generate
        self.a, self.b = 0, 1  #starting math positions: a=0, b=1
        self.count = 0  #tracks how many numbers have been printed

    def __iter__(self):
        return self

    def __next__(self):
        #checks if we hit the limit then raises stopiteration
        if self.count >= self.n:
            raise StopIteration
        #saves the current number to return to the loop
        fib_num = self.a
        #the fibonacci math(the new a becomes the old b while the new b becomes the sum of old a and b)
        self.a, self.b = self.b, self.a + self.b
        #tracks loop and return the number
        self.count += 1
        return fib_num

fib = Fibonacci(10)

for num in fib:
    print(num)
````

Create classes `Rectangle`, `Circle`, and `Triangle`, each with an `area()` method.  
- Store instances of these classes in a list.  
- Write code that calculates the total area without checking object types.  

````python
import math

#classes rectangle, circle and triangle with area method
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    def area(self):
        return self.width * self.height
        
class Circle:
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return math.pi * (self.radius ** 2)

class Triangle:
    def __init__(self,a,b,c):
        self.a = a
        self.b = b
        self.c = c
        
    def area(self):
        s = math.sqrt(self.a + self.b + self.c) / 2
        return math.sqrt(s * (s - self.a) * (s - self.b) * (s - self.c))

#instances of classes in a list  
shapeList = [
        Rectangle(5,4),
        Circle(8),
        Triangle(1,2,3),
        Rectangle(2,6),
        Circle(4),
        Triangle(3,1,4)
]
for shape in shapeList:
    print(f"Area: {shape.area():.2f}")
````

````python
#Library Management System

#OOP and Inheritance
#Polymorphism
#Iterators
#Scope

#base class Book with attributes title, author, year
class Book:
    def __init__(self,title,author,year):
        self.title = title
        self.author = author
        self.year = year
        self._is_borrowed = False  #protected: tracks availability

#method get_description
    def get_description(self):
        return f"{self.title} by {self.author} ({self.year})"

#subclass ebook with attributes filesize and format
class Ebook(Book):
    def __init__(self, title, author, year, file_size, format):
        super().__init__(title, author, year)
        self.file_size = file_size
        self.format = format

#method get_description
    def get_description(self):
        return f"{super().get_description()} [Ebook, Format: {self.file_size}MB, Size: {self.format}]"
    

#subclass PrintedBook with attributes weight and dimensions
class PrintedBook(Book):
    def __init__(self, title, author, year, weight, dimensions):
        super().__init__(title, author, year)
        self.weight = weight
        self.dimensions = dimensions

#method get_description
    def get_description(self):
        return f"{super().get_description()} [Printed, Weight: {self.weight}g, Dimensions: {self.dimensions}cm]"

````

````python
#base class Member with name and member_id
class Member:
    total_books_borrowed = 0  #class-level variable shared across all member instances

    def __init__(self,name,member_id):
        self.name = name
        self.member_id = member_id
        self._borrowed_book = []  #protected: list of book objects
  
    #implements an iterator to iterate over all books
    def __iter__(self):
        return MemberIterator(self._borrowed_book)
    
#subclass studentmember with gradelevel
class StudentMember(Member):
    BORROW_LIMIT = 3  

    def __init__(self, name, member_id, grade_level):
        super().__init__(name, member_id)
        self.grade_level = grade_level

#method borrow_book which checks how many books a student has compared with the borrow
#limit and returns false.
    def borrow_book(self, book):
        if len(self._borrowed_book) >= self.BORROW_LIMIT:
            print(f"{self.name} has reached the student borrow limit ({self.BORROW_LIMIT}).")
            return False

        #tracks which book is borrowed
        if book._is_borrowed:
            print(f"{book.title} is already borrowed")
            return False

        #tracks available book and appends that while counting the number of books borrowed by each student 
        book._is_borrowed = True
        self._borrowed_book.append(book)
        Member.total_books_borrowed += 1
        print(f"{self.name} (student) borrowed '{book.title}'.")
        return True
    
#subclass teachermember with department
class TeacherMember(Member):
    BORROW_LIMIT = 10

    def __init__(self, name, member_id, department):
        super().__init__(name, member_id)
        self.department = department

#method borrow_book which checks how many books a teacher has compared with the borrow
#limit and returns false.
    def borrow_book(self, book):
            if len(self._borrowed_book) >= self.BORROW_LIMIT:
                print(f"{self.name} has reached the teacher borrow limit ({self.BORROW_LIMIT}).")
                return False

            #tracks which book is borrowed
            if book._is_borrowed:
                print(f"{book.title} is already borrowed")
                return False

            #tracks available book and appends that while counting the number of books borrowed by each teacher 
            book._is_borrowed = True
            self._borrowed_book.append(book)
            Member.total_books_borrowed += 1
            print(f"{self.name} (teacher) borrowed '{book.title}'.")
            return True
````

````python
#implements an iterator for the member class to iterate over all books
class MemberIterator:
    def __init__(self,borrowed_book):
        self._borrowed_book = borrowed_book  #keep a reference to the list of members borrowed books list
        self._index = 0  #tracks current position in that list but starts at the beginning

    def __next__(self):
        #checks if we're within the valid index
        if self._index >= len(self._borrowed_book):
            raise StopIteration

        #grabs the book at current position, move one step forward so next value returns the next book
        book = self._borrowed_book[self._index]
        self._index += 1
        return book
````

````python
#implements an iterator for the library class to iterate over all books
class LibraryIterator:
    def __init__(self,books):
        self._books = books  #keep a reference to the list of books we're interacting over
        self._index = 0  #tracks current position in that list but starts at the beginning

    def __next__(self):
        #checks if we're within the valid index
        if self._index >= len(self._books):
            raise StopIteration

        #grabs the book at current position, move one step forward so next value returns the next book
        book = self._books[self._index]
        self._index += 1
        return book
````

````python
#library class

class Library:
    def __init__(self,name):
        self.__library_name = name  #private attribute
        self._books = []  #protected: holds every book object added to the library

    #getter and setter method to access protected and private attribute
    def get_library_name(self):
        return self.__library_name

    def set_library_name(self, new_name):
        if new_name.strip():  #remove spaces and checks for empty values
            self.__library_name = new_name
        else:
            print("Library name cannot be empty")

    #a function that appends to the internal list
    def add_book(self,book):
        self._books.append(book)
        print(f"Added '{book.title}' to the library")

#implements an iterator to iterate over all books
    def __iter__(self):
        return LibraryIterator(self._books)

    #tracks avaiable vs borrowed books
    def available_books(self):
        return [book for book in self._books if not book._is_borrowed]

    def borrowed_books(self):
            return [book for book in self._books if book._is_borrowed]

    #search by title/author and converts it to lowercase
    def search_by_title(self,keyword):
        return [book for book in self._books if keyword.lower() in book.title.lower()]

    def search_by_author(self,keyword):
            return [book for book in self._books if keyword.lower() in book.author.lower()]
    
````

````python
#Test Script

#Adding different books to the library

lib = Library("De-Cheson Library")

book1 = Ebook("Dune", "Frank Herbert", 1965, 3.2, "EPUB")
book2 = PrintedBook("1984", "George Orwell", 1949, 320, "20x13x2 cm")
book3 = PrintedBook("Animal Farm", "George Orwell", 1945, 150, "18x11x1 cm")
book4 = Ebook("Brave New World", "Aldous Huxley", 1932, 2.1, "PDF")

for b in [book1, book2, book3, book4]:
    lib.add_book(b)
````
