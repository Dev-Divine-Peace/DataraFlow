```python
#prints the multiplication table from 1-12
number = int(input("Select a number: "))
for i in range(1, 13):
    print(number, "x", i, "=", number * i)
```

```python
#Calculate the factorial of a number using for loop
def factorial(n):
    factorize = 1
    for i in range(1, n+1):
        factorize *= i
    return factorize
print(factorial(6))
```

```python
#program that finds the largest number in a given list
numbers = [17, 8, 2, 30, 10, 23]
print(max(numbers))
```

```python
#Counts how many vowels in a given string
letters = input("Type a word: ")
count = 0
for letter in letters:
    if letter.lower() in "aeiou":
        count += 1
print("Number of vowels: ", count)
```

```python
#Function which takes 2 functions and returns its sum, difference and product.
def sum(x,y):
    return x + y
def difference(x,y):
    return x - y
def product(x,y):
    return x * y

num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

print("The sum is ", sum(num1,num2))
print("The difference is ", difference(num1,num2))
print("The product is ", product(num1,num2))
```

```python
#Reverses a string without using slicing
def str_reverse(a):
    str_reversed = " "
    for char in a:
        str_reversed = char + str_reversed
    return str_reversed
print (str_reverse("Divine-Peace"))
```

```python
#Generats the first 10 numbers of the fibbonaci sequence
def fibonacci_seq(n):
    a,b = 0,1
    for _ in range(n):
        print(a)
        a,b = b, b + a
fibonacci_seq(10)
```

## Assignments

1. **Student Grades System**  
   Write a program to accept marks for 5 subjects, calculate the average, and display the grade based on:  
   - A: 80–100  
   - B: 60–79  
   - C: 40–59  
   - F: < 40  

2. **Simple Calculator**  
   Build a calculator program that can perform addition, subtraction, multiplication, and division based on user input.

3. **Shopping Cart**  
   Create a program where a user can add items to a shopping cart (dictionary with item names and prices), then display the total cost.

4. **Guessing Game**  
   Write a program where the computer randomly selects a number between 1–20, and the user must guess it. Give hints if the guess is too high or too low.

5. **File Handling**  
   Write a program that accepts a list of student names from the user, saves them in a text file, and then reads and displays them.


```python
#Student grade system

#function to calculate grades from the average marks
def calculate_grade(marks):
    total = 0

    for mark in marks:
        total+= mark
#calculates the average by the sum of all marks divided by the number of marks in the list.
    average = total / len(marks)

    if average >= 80:
        grade = "A"
    elif average >= 60:
        grade = "B"
    elif average >= 40:
        grade = "C"
    else:
        grade = "F"
    return average, grade

#function to store the marks from 5 subjects and display the average marks and grade.
def studentGrade():
    subjects = ["Maths", "English", "Chemistry", "Biology", "Physics"]
    marks = []

    for subject in subjects:
        mark = float(input(f"Enter marks for {subject}: "))
        marks.append(mark)

    average, grade = calculate_grade(marks)
    print(f"Average Marks: {average:.2f}")
    print(f"Grade: {grade}")

studentGrade()
```

```python
#Simple calculator
#functions for addition, subtraction, multiplication and division of two numbers choosen by the user
def addition(a,b):
    return a + b
def subtraction(a,b):
    return a - b
def multiplication(a,b):
    return a * b
def division(a,b):
    return a / b
print("Choose an operation: +, -, *, /")
operation = input("Operation: ")
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
if operation == "+":
    print("The result is ", addition(num1,num2))
elif operation == "-":
    print("The result is ", subtraction(num1,num2))
elif operation == "*":
    print("The result is ", multiplication(num1,num2))
elif operation == "/":
    print("The result is ", division(num1,num2))
else:
    print("Invalid Operation")
```

```python
#Shopping cart
shopping_cart = {}
while True:
    print("\nShopping Cart Menu")
    print("1. Add item")
    print("2. View cart and total")
    print("3. Exit")
    choice = input("Enter your choice (1-3): ")
    if choice == "1":
        item = input("Enter the item name: ")
        price = float(input("Enter the item price: "))
        shopping_cart[item] = price
        print(f"{item} added to cart!")
    elif choice == "2":
        if not shopping_cart:
            print("Your cart is empty.")
        else:
            print("\nItems in your cart:")
            total = 0
            for item, price in shopping_cart.items():
                print(f"- {item}: N{price:.2f}")
                total += price
            print(f"Total cost: N{total:.2f}")
    elif choice == "3":
        print("Thanks for shopping!")
        break
    else:
        print("Invalid choice, please try again.")
```

```python
#Generate a random number
import random
number_to_guess = random.randint(1,20)
while True:
    try:
        guess = int(input("Guess the number between 1 and 20: "))
        if guess < number_to_guess:
            print("Too low")
        elif guess > number_to_guess:
            print("Too high")
        else:
            print("Well done")
            break
    except ValueError:
        print("Please enter a valid number")
```

```python
#File Handling
studentData = []
num = int(input("How many students? "))
for i in range(num):
    name = input(f"Enter name of student {i+1}: ")
    studentData.append(name)
with open("students.txt", "w") as file:
    for name in studentData:
        file.write(name + "\n")
print("Student names saved to students.txt")
print("\nReading from file...")
with open("students.txt", "r") as file:
    content = file.readlines()

for i, name in enumerate(content, start=1):
    print(f"{i}. {name.strip()}")
```
