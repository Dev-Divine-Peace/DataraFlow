```python
#Task 1: The square root of 144

import math
x = math.sqrt(144)
print(x)
```

```python
#Task 2: A program that prints current date and time using datetime module

import datetime
currentDate = datetime.datetime.now()
print(currentDate)
```

```python
#Task 3: Convert Python dictionary into a JSON string 

import json
pyDictionary = {"name":"Divine-Peace", "age":22, "hobby":"Eating"}
y = json.dumps(pyDictionary)  #dumps: convert to a JSON string
print(y)
```

```python
#Task 4: Parse a JSON string into Python dictionary

import json
jsDict = '{"name":"Divine-Peace", "age":22, "hobby":"Eating"}'
dict = json.loads(jsDict)
print(dict["name"])
```

```python
#Task 5: Calculate the catorial of 6

import math
result = math.factorial(6)
print(result)
```

```python
#Task 6: Program that takes a user's birth year and calculates their age

import datetime

birth_year = int(input("Enter your birth year: "))
current_year = datetime.datetime.now().year
age= current_year - birth_year
print("You are", age, "years old.")
```

```python
#Task 7: A try/except block that handles division by zero

try:
    firstNo = float(input("Enter numerator: "))
    secondNo = float(input("Enter denominator: "))
    result = firstNo/secondNo
    print("Result:", result)
except ZeroDivisionError:
    print("Error: cannot divide by zero")
```

```python
#Task 9: Display today's data in the format DD-MM-YYYY

from datetime import date
today = date.today()
todayDate = today.strftime("%d-%m-%Y")
print(todayDate)
```

```python
#Task 10: Program that prompts user for a number and prints if it is positive, negative or zero

num = float(input("Enter a number: "))

if num > 0:
    print("Positive")
elif num < 0:
    print("Negative")
else:
    print("Zero")
```
