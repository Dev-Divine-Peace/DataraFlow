```python
"""Assignment 1: A program that stores a list of student records (name,age,grade) into a JSON 
file and then reads the file back into a python object."""

#Step 1: Import json module
import json

#Step 2: created a list of student records (each student is represented in a dictionary)
students = [
    {"name":"Divine", "age":22, "grade":"A"},
    {"name":"Joshua", "age":19, "grade":"B"},
    {"name":"Prosper", "age":21, "grade": "A+"}
]

#Step 3:storing the list into a JSON file specifying the number of spaces for each indentation level
with open("students.json", "w") as file:
    json.dump(students,file, indent=4)

#Step 4: Read the file back into a python object
with open("students.json", "r") as file:
    loaded_student = json.load(file)

#Step 5: print the file
print(loaded_student)
```

```python
"""Assignment 2: A program that asks the user 5 random math questions (addition, 
subtraction, multiplication and division) and keeps score of correct answers"""

#Step 1: Import random module 
import random

#Step 2: store the operators required in a variable and assigning a variable called
# score with value 0 for keeping track of correct answers
operators = ["+", "-", "*", "/"]
score = 0

for i in range(5):
    # Step 3: generate 2 random numbers
    num1 = random.randint(1,15)
    num2 = random.randint(1,15)
    operator = random.choice(operators)  #randomly choose operator

    #Step 4: Handles division by making sure num2 isn't zero
    if operator == "/":
        num2 = random.randint(1,10)

    #Step 5: creates the question string
    question = f"{num1} {operator} {num2}"  
    
    #Step 6: Calculates the correct answer
    if operator == "+":
        correct_answer = num1 + num2
    elif operator == "-":
        correct_answer = num1 - num2
    elif operator == "*":
        correct_answer = num1 * num2
    elif operator == "/":
        correct_answer = num1 / num2

    #Step 7: Asks the user
    user_input = float(input(f"Question {i + 1}: {question} = "))

    #Step 8: Checks if the answer is correct
    if round(user_input, 2) == round(correct_answer,2):
        print("Correct Answer!")
        score += 1
    else:
        print(f"Wrong. Correct answer is {correct_answer}")

#Step 9: Print final score
print(f"Your final score is {score}/5")       
                
```

```python
"""Assignment 3: A program that asks the user for a number and divides 100 by that
number using try/except to handle invalid input and division by zero"""

try:
    #Step 1: Asks for input
    number = float(input("Enter a number to divide 100 by: "))

    #Step 2: Divide 100 by user input
    result = 100 / number
    print(f"100 divided by {number} is {result}")

    #Step 3: Handle invalid number e.g text
except ValueError:
    print("Invalid input! Please enter a numeric value.")

    #Step 4: Handles division by zero
except ZeroDivisionError:
    print("You can't divide by zero!")
```

```python
"""Assignment 4: A program that takes a user's birthday as input and tells them
how many days are left until their next birthday"""

#Step 1: Import module
import datetime

#Step 2: Ask user for birthday
day = int(input('Enter your birth day (1-31): '))
month = int(input('Enter your birth month (1-12):'))

#Step 3: Get today's date
today = datetime.date.today()
year = today.year

#Step 4: Create birthday date for this year
birthday = datetime.date(year,month,day)

#Step 5: If birthday date has passed, use next year
if birthday < today:
    birthday = datetime.date(year + 1, month,day)

#Step 6: Calculate days difference
days_left = (birthday - today).days
print(f"Your next birthday is in {days_left} days.")
```

```python
#9. A program that asks the user for two numbers and prints their greatest common divisor

import math

def gcdCalculate():
    num1 = int(input("Enter the first number: "))
    num2 = int(input("Enter the second number: "))

    gcd = math.gcd(num1,num2) #finds the greatest common divisor
    print(f"The GCD of {num1} and {num2} is {gcd}")

gcdCalculate()
```
