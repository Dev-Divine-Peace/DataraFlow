````python
#Task 1: A program to test element-wise for NaN of a given array

#array with a NaN value
arr = np.array([1, 2, np.nan, 4, np.nan, 6])
print("Original Array:")
print(arr)

#Test element-wise  for NaN
result = np.isnan(arr)
print("\nElement-wise NaN Check:")
print(result)
````

````python
#Task 2: A program testing element-wise for complex number, real number of a given array

#Array with different data types
arr = np.array([1, 2+3j, 4.5, 3-2j, 5])
print("Original Array:")
print(arr)

#Element-wise test for complex numbers
is_complex = np.iscomplex(arr)
print("\nElement-wise Complex Check:")
print(is_complex)

#Element-wise test for real numbers
is_real = np.isreal(arr)
print("\nElement-wise Real Check:")
print(is_real)

#Test whether scalar_num and array_num is a scalar type or not
scalar_num = 5
array_num = np.array([1, 2, 3])

is_scalar_1 = np.isscalar(scalar_num)
print("\nIs 5 a scalar?")
print(is_scalar_1)

is_scalar_2 = np.isscalar(array_num)
print("\nIs [1,2,3] a scalar?")
print(is_scalar_2)
````

````python
#Task 3: A program to test whether two arrays are element-wise equal within a tolerance

#Create two arrays that are close but not equal
arr1 = np.array([1.0, 2.001, 3.0, 4.0005])
arr2 = np.array([1.0, 2.0, 3.0001, 4.0])
print("Array 1:", arr1)
print("Array 2:", arr2)

#Test if the two arrays are equal within a tolerance
result = np.allclose(arr1, arr2, atol=0.001)
print("\nAre the arrays element-wise equal within a tolerance?")
print(result)
````

````python
#Task 4: A program to create an element-wise comparison of two given arrays

#Two given arrays
arr1 = np.array([10,20,30,40,50])
arr2 = np.array([40,50,60,60,50])
print("Array 1:", arr1)
print("Array 2:", arr2)

#A greater element-wise comparison
greater = np.greater(arr1, arr2)
print("\nElement-wise 'greater' comparison (arr1 > arr2):")
print(greater)

#A greater_equal element-wise comparison
greater_equal = np.greater_equal(arr1, arr2)
print("\nElement-wise 'greater_equal' comparison (arr1 >= arr2):")
print(greater_equal)

#A less element-wise comparison
less = np.less(arr1, arr2)
print("\nElement-wise 'less' comparison (arr1 < arr2):")
print(less)

#A less_equal element-wise comparison
less_equal = np.less_equal(arr1, arr2)
print("\nElement-wise 'less_equal' comparison (arr1 <= arr2):")
print(less_equal)
````

````python
#Task 5: A program to cretae an element-wise comparison of two given arrays

#Create two arrays
arr1 = np.array([10,20,30.0001,40])
arr2 = np.array([40,50,60,60])
print("Array 1:", arr1)
print("Array 2:", arr2)

#An equal element-wise comparison
equal = np.equal(arr1, arr2)
print("\nElement-wise Exact Equality (arr1 == arr2):")
print(equal)

#An equal within a tolerance element-wise comparison
close = np.isclose(arr1, arr2, atol=20)
print("\nElement-wise Equality Within a Tolerance (|a - b| <= 20):")
print(close)
````

### SQL Task 2 · Filtering & NULLs
For **London** (`week06.london_2014`), return **one row** with three counts:
1. `no_event_days`: days with **no event recorded**
2. `rain_days`: days whose event mentions **Rain** anywhere (e.g. `Rain`, `Fog-Rain`, `Rain-Thunderstorm`)
3. `wet_but_no_event_days`: days with **no event recorded** but `Precipitationmm > 0`

⚠️ The events column name starts with a **space**: `[ Events]`. And remember, `= NULL` never matches.
*Revisit: §1.2, §3.2, §3.3.*

````python
SQL_02 = """
        SELECT
            SUM((CASE WHEN ([ EVENTS] IS NULL) THEN 1 ELSE 0 END)) AS no_event_days,
            SUM((CASE WHEN ([ Events] LIKE '%Rain%') THEN 1 ELSE 0 END)) AS rain_days,
            SUM((CASE WHEN ([ Events] IS NULL) AND (Precipitationmm > 0) THEN 1 ELSE 0 END)) AS wet_but_no_event_days

        FROM week06.london_2014
"""
answer("s02", SQL_02)
````

### SQL Task 7 · UNION ALL
Stack the **six** city weather tables in `week06` and compute each city's **average `Mean TemperatureC`**
for 2014, **rounded to 1 decimal place**. Use exactly these city labels:
`'Beijing'`, `'Brasilia'`, `'Cape Town'`, `'Delhi'`, `'London'`, `'Moscow'`.

**Return:** `city`, `avg_mean_temp_c`, sorted by `city` (A→Z).
⚠️ `Mean TemperatureC` is stored as a whole number, so watch the integer-division trap. *Revisit: §8.1–8.2.*

````python
# Query the six different cities' table names

run_query("""
        SELECT TABLE_SCHEMA + '.' + TABLE_NAME
        FROM INFORMATION_SCHEMA.TABLES
        WHERE TABLE_SCHEMA = 'week06'
""")
````

````python
SQL_07 = """
SELECT city, ROUND(AVG(CAST(mean_temp_c AS FLOAT)), 1) AS avg_mean_temp_c
FROM(
        SELECT 'Beijing' AS city,     [Mean TemperatureC] AS mean_temp_c FROM week06.beijing_pek_2014
        UNION ALL SELECT 'Brasilia',  [Mean TemperatureC] FROM week06.brasilia_bsb_2014
        UNION ALL SELECT 'Cape Town', [Mean TemperatureC] FROM week06.capetown_cpt_2014
        UNION ALL SELECT 'Delhi',     [Mean TemperatureC] FROM week06.delhi_del_2014
        UNION ALL SELECT 'London',    [Mean TemperatureC] FROM week06.london_2014
        UNION ALL SELECT 'Moscow',    [Mean TemperatureC] FROM week06.moscow_svo_2014
) AS combined_weather
GROUP BY city
ORDER BY city                                
"""
answer("s07", SQL_07)
````

### SQL Task 8 · CTE + window ranking (top-1 per group)
For **London**, find the **hottest day of each month** (highest `Max TemperatureC`). If two days tie,
take the **earlier** date.

**Return:** `month_no`, `day` (a real **date**, convert `GMT` with `TRY_CONVERT(date, GMT)`), `max_temp`,
sorted by `month_no`. *Revisit: §10, §11.3.*

````python
SQL_08 = """
WITH RankedTemperatures AS (
    SELECT
        TRY_CONVERT(date, GMT) AS day,
        MONTH(TRY_CONVERT(date, GMT)) AS month_no,
        [Max TemperatureC] AS max_temp,
        ROW_NUMBER() OVER (
            PARTITION BY MONTH(TRY_CONVERT(date, GMT))
            ORDER BY [Max TemperatureC] DESC, TRY_CONVERT(date, GMT) ASC
        ) AS rn
    FROM week06.london_2014
)
SELECT
    month_no,
    day,
    max_temp
FROM RankedTemperatures
WHERE rn = 1
ORDER BY month_no;
"""
answer("s08", SQL_08)
````

### SQL Task 9 · Running total
From `week09.comtrade_milk_uk_monthly_14`, compute the UK's **total milk exports per month** (exclude `'World'`)
and a **year-to-date running total**.

**Return:** `Period`, `exports_usd`, `ytd_exports_usd`, sorted by `Period`. *Revisit: §11.4–11.5.*

````python
SQL_09 = """
WITH MonthlyExports AS (
    SELECT
        Period,
        SUM([Trade Value (US$)]) AS exports_usd
    FROM week09.comtrade_milk_uk_monthly_14
    WHERE partner != 'World'
      AND [Trade Flow] = 'Exports'
    GROUP BY Period
)

SELECT
    Period,
    exports_usd,
    SUM(exports_usd) OVER (ORDER BY Period) AS ytd_exports_usd
FROM MonthlyExports
ORDER BY Period
"""
answer("s09", SQL_09)
````
